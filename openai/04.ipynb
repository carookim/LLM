## RAG 시스템

#### LLM의 한계

- 최신 정보 부족
    - 학습 시점 이후의 사건·데이터는 알지 못함.
    - 실시간 정보가 필요한 경우 외부 연동 필요.

- 사실 검증 어려움
    - 출력 내용이 항상 정확하지 않음.
    - 잘못된 정보나 허구도 실제처럼 제시할 수 있음.

- 창의성 한계
    - 기존 데이터 패턴을 기반으로 답변 생성.
    - 완전히 새로운 지식이나 독창적 발명은 어려움.

- 맥락·상식 오류 가능
    - 긴 대화나 복잡한 논리에서 일관성 저하.
    - 상식적 판단이 사람 수준과 다를 수 있음.

- 편향·윤리적 문제
    - 학습 데이터의 편향이 답변에 반영될 수 있음.
    - 민감한 주제에서 부적절한 답변 가능.

```python
# RAG
# 우리회사의 복지제도는 ?
# LLM은 학습데이터에 없는 최신 / 특정 정보를 모른다.
    # LLM은 데이터 업데이트가 중요하겠구나.

# RAG 해결책  (Tetrieval Augmented Generation) = 검색 + 생성
# 1. 회사문서에서 관련정보를 검색
# 2. 검색된 정보를 LLM에게 context로 제공
# 3. LLm이 context 기반으로 정확한 답변 생성

# [준비 단계]
# 문서들 -> 청크 분할 -> 벡터변환 -> 벡터DB저장

# [쿼리 단계]
# 질문 -> 벡터변환 -> 유사도검색 -> 상위 k개 선택 -> context + 질문 -> LLM -> 답변
```
