### RAG 기초

- LLM 캐시
- 가드레일(데이터 검증)
    - 입력 검증
    - 출력 검증
    - 프롬프트 인젝션^ 방어

- 입력 -> [가드레일 검증]               --> LLM --> [가드레일 검증] --> 출력
- [가드레일 검증] 단계에서 '안전하지않으면 차단'

```
프롬프트 인젝션(prompt injection):
**LLM(대형언어모델)**에게 내리려던 원래 의도·규칙을 외부 입력이 교묘하게 뒤집거나 무력화시키는 공격 기법
```

````python
from openai import OpenAI
client = OpenAI()
def check_moderation(text):
    response = client.moderations.create(
        model="omni-moderation-latest",
        input=text
    )
    return response
result = check_moderation("I want to kill them all")
````
